# Portfolio Analytics: Maximum Sharpe Ratio (Tangency) Portfolio

Runs `portfolio.maximum_sharpe_ratio_portfolio` -- SLSQP-optimised weights maximising the annualised Sharpe ratio subject to weights summing to 1 (`engine/portfolio_optimisation.py`). Requires a free-tier API key: https://www.pyvar.com#get-api-key

```python
%load_ext pyvar_jupyter
%pyvar_key eyJ...  # replace with your own key, or set PYVAR_API_KEY before starting the kernel
```

## Four assets, expected returns and a covariance matrix

Annualised figures, a synthetic but plausible cross-asset set (equity-like, higher-vol equity-like, bond-like, balanced) -- in practice these come from a factor model or historical estimation, not hand-typed like this.

```python
mean_returns = [0.08, 0.11, 0.06, 0.09]
cov_matrix = [
    [0.0400, 0.0150, 0.0050, 0.0100],
    [0.0150, 0.0900, 0.0100, 0.0200],
    [0.0050, 0.0100, 0.0225, 0.0060],
    [0.0100, 0.0200, 0.0060, 0.0625],
]
risk_free = 0.03
```

## Run it (default: long-only, weights bounded to [0, 1])

```python
result = %pyvar portfolio.maximum_sharpe_ratio_portfolio mean_returns={mean_returns} cov_matrix={cov_matrix} risk_free={risk_free} periods_per_year=1
result
```

The optimiser lands on roughly 30% / 20% / 28% / 22% across the four assets (`weights`), an 8.26% expected return, 13.99% volatility, and a Sharpe ratio of 0.376 -- with `success: true` confirming SLSQP actually converged rather than returning a fallback. Note asset 2 (11% expected return, the highest of the four) doesn't dominate the allocation -- its 30% standalone variance is the highest too, so the optimiser trades some of that expected return away for the diversification the lower-correlation assets provide.
