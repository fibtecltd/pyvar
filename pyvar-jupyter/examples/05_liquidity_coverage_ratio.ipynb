# Liquidity Risk: Basel III Liquidity Coverage Ratio (LCR)

Runs `liquidity_risk.liquidity_coverage_ratio_lcr` -- `LCR = HQLA / net 30-day outflows`, with inflows capped at 75% of gross outflows (BCBS 238 §69, `engine/liquidity_ratios.py`) so net outflows can never fall below 25% of gross. Requires a free-tier API key: https://www.pyvar.com#get-api-key

```python
%load_ext pyvar_jupyter
%pyvar_key eyJ...  # replace with your own key, or set PYVAR_API_KEY before starting the kernel
```

## A 30-day liquidity position

£120m of post-haircut high-quality liquid assets, £150m of stressed 30-day outflows, and £40m of contractual inflows -- comfortably under the 75% inflow cap, so nothing gets floored here.

```python
hqla = 120_000_000
gross_outflows = 150_000_000
gross_inflows = 40_000_000
hqla, gross_outflows, gross_inflows
```

## Run it

```python
result = %pyvar liquidity_risk.liquidity_coverage_ratio_lcr hqla={hqla} gross_outflows={gross_outflows} gross_inflows={gross_inflows}
result
```

`net_outflows` comes out to £110,000,000 (£150m − £40m of *uncapped* inflows, since £40m sits below the £112.5m cap), giving `lcr: 1.090909` -- 109.1%, above the 100% regulatory minimum, so `compliant: true`. Push `gross_inflows` up past £112.5m in this same cell and watch `capped_inflows` stop moving with it -- a quick, concrete way to see the 75% cap actually bind.
