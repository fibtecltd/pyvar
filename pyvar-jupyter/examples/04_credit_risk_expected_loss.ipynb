# Credit Risk: Expected Loss (EL = PD × LGD × EAD)

Runs `credit_risk.expected_loss_el_computation` -- the standard IRB building block for provisioning and pricing a single exposure (`engine/credit_pd_lgd.py`'s own docstring: `EL = PD * LGD * EAD`). Requires a free-tier API key: https://www.pyvar.com#get-api-key

```python
%load_ext pyvar_jupyter
%pyvar_key eyJ...  # replace with your own key, or set PYVAR_API_KEY before starting the kernel
```

## A single corporate exposure

A 2% one-year probability of default, 45% loss given default (a common unsecured-senior assumption), and a £5,000,000 exposure at default -- the three inputs any IRB expected-loss figure reduces to.

```python
pd = 0.02
lgd = 0.45
ead = 5_000_000
pd, lgd, ead
```

## Run it

`expected_loss_el_computation` is synchronous -- no polling, direct JSON response, same as 384 of pyvar's 385 functions.

```python
result = %pyvar credit_risk.expected_loss_el_computation pd={pd} lgd={lgd} ead={ead}
result
```

With these inputs, `el` comes out to exactly `45,000.0` (£5,000,000 × 0.02 × 0.45) and `el_rate` to `0.009` -- 0.9% of exposure, the loss rate a provisioning model would carry for this facility before any stage-2/stage-3 IFRS 9 overlay. Because EL is just a product of the three inputs, this is a useful sanity-check function before trusting a more elaborate portfolio-level credit model: if a portfolio aggregate doesn't reconcile to the sum of exposures run through this formula, something upstream is wrong.
