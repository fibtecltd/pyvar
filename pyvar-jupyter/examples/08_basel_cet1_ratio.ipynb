# Regulatory: Basel III CET1 Capital Ratio

Runs `regulatory.basel_iii_cet1_ratio` -- `CET1 ratio = CET1 capital / RWA`, checked against the 4.5% Basel III §50 minimum (`engine/reg_basel_capital.py`). Requires a free-tier API key: https://www.pyvar.com#get-api-key

```python
%load_ext pyvar_jupyter
%pyvar_key eyJ...  # replace with your own key, or set PYVAR_API_KEY before starting the kernel
```

## A bank-scale capital position

£8.5bn of Common Equity Tier 1 capital against £62bn of risk-weighted assets -- illustrative figures at the scale a mid-size bank's Pillar 3 disclosure would report.

```python
cet1_capital = 8_500_000_000
risk_weighted_assets = 62_000_000_000
```

## Run it

```python
result = %pyvar regulatory.basel_iii_cet1_ratio cet1_capital={cet1_capital} risk_weighted_assets={risk_weighted_assets}
result
```

`cet1_ratio` comes out to `0.13709677` -- 13.71%, well above the 4.5% `minimum`, with `surplus: 0.09209677` (9.21 percentage points of headroom) and `compliant: true`. That headroom is before the capital-conservation buffer, any countercyclical buffer, or a Pillar 2A/2B add-on -- each of those is its own function in the same `regulatory` namespace (`combined_buffer_requirement`, `countercyclical_capital_buffer`, `pillar_2a_capital`), stacked on top of this minimum rather than replacing it.
