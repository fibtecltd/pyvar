# ALM: Economic Value of Equity (EVE) Sensitivity Under IRRBB's Six Standard Shocks

Runs `alm.eve_sensitivity_analysis` -- ΔEVE under each of BCBS d368's six prescribed interest-rate shock scenarios, plus the worst case: the headline IRRBB supervisory metric (`engine/alm_nii_eve.py`). Requires a free-tier API key: https://www.pyvar.com#get-api-key

```python
%load_ext pyvar_jupyter
%pyvar_key eyJ...  # replace with your own key, or set PYVAR_API_KEY before starting the kernel
```

## A net cash-flow ladder across six tenor buckets

Net (asset minus liability) cash flow per bucket, the bucket's time in years, and the base zero rate at that tenor -- the three parallel arrays every IRRBB EVE calculation reduces to, whatever the underlying instrument mix.

```python
times = [0.25, 1.0, 2.0, 5.0, 10.0, 20.0]
net_cashflows = [15_000_000, 25_000_000, -10_000_000, 40_000_000, -20_000_000, 30_000_000]
base_rates = [0.035, 0.037, 0.039, 0.042, 0.044, 0.045]
```

## Run it (default shock magnitudes: 200bp parallel, 300bp short, 150bp long)

```python
result = %pyvar alm.eve_sensitivity_analysis times={times} net_cashflows={net_cashflows} base_rates={base_rates}
result
```

`base_eve` comes out to roughly £61.45m. Of the six shocks, `parallel_up` produces the largest fall (−£4.96m), which is also `worst_case` here -- a longer-duration liability book (net cash flows further out on the curve) would instead see `parallel_down` or `steepener` dominate; which of the six binds is precisely why the framework tests all six rather than assuming parallel risk is always the binding one. The IRRBB supervisory outlier test compares this worst-case ΔEVE against a fixed percentage of Tier 1 capital -- a threshold this function deliberately doesn't apply, since that comparison needs a real capital figure, not a synthetic one.
